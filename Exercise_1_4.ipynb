• If the number is greater than or equal to 150, then skip it.
• If the number is greater than 500, then stop the loop.

Example input:

```python
n = [12, 75, 150, 182, 145, 525, 50]
for int in n:
    if int > 500:
        break
    if int >= 150:
        continue
    if int % 5 == 0:
        print(int)
```

```python
def count_digits(n):
    count = 0
    while n > 0:
        n = n // 10
        count = count + 1
    return count

n = 75869
print(count_digits(n))
```
